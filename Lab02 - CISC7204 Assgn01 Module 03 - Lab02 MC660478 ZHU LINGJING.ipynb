# Task 3 - GroupBy and Pivot Tables

Group the parameters "GPU", "CPU_core" and "Price" to make a pivot table and visualize this connection using the pcolor plot.


````python
# Write your code below and press Shift+Enter to execute
# Create the group
# Select the three columns for grouping analysis
df_gptest = df[['GPU', 'CPU_core', 'Price']]

# Group by GPU and CPU_core, calculate the mean price for each combination
# as_index=False keeps GPU and CPU_core as regular columns instead of becoming the index
# numeric_only=True ensures only numeric columns are aggregated (avoids errors)
grouped_test1 = df_gptest.groupby(['GPU', 'CPU_core'], as_index=False).mean(numeric_only=True)
print(grouped_test1)
````

<summary>Click here for Solution</summary>


</details>


````python
# Write your code below and press Shift+Enter to execute
# Create the Pivot table
# Convert the grouped data into a pivot table for easier visualization
# index='GPU' puts GPU values as rows
# columns='CPU_core' puts CPU_core values as columns
# The cell values are the mean Price for each GPU-CPU_core combination
grouped_pivot = grouped_test1.pivot(index='GPU', columns='CPU_core')
print(grouped_pivot)
````

<details>
<summary>Click here for Solution</summary>


</details>


````python
# Write your code below and press Shift+Enter to execute
# Create the Plot
# Create a pseudocolor plot (heatmap) of the pivot table
fig, ax = plt.subplots()
im = ax.pcolor(grouped_pivot, cmap='RdBu')

# Get label names from the pivot table MultiIndex structure
# columns.levels[1] gets the CPU_core values
row_labels = grouped_pivot.columns.levels[1]
# index gets the GPU values
col_labels = grouped_pivot.index

# Move ticks and labels to the center of each cell
ax.set_xticks(np.arange(grouped_pivot.shape[1]) + 0.5, minor=False)
ax.set_yticks(np.arange(grouped_pivot.shape[0]) + 0.5, minor=False)

# Insert the actual labels on the axes
ax.set_xticklabels(row_labels, minor=False)
ax.set_yticklabels(col_labels, minor=False)

# Add a colorbar to show the value scale
fig.colorbar(im)
plt.show()
````

# Task 4 - Pearson Correlation and p-values

Use the `scipy.stats.pearsonr()` function to evaluate the Pearson Coefficient and the p-values for each parameter tested above. This will help you determine the parameters most likely to have a strong effect on the price of the laptops.


````python
# Write your code below and press Shift+Enter to execute
# Loop through each parameter and calculate Pearson correlation with Price
# stats.pearsonr() returns both the correlation coefficient and the p-value
# p-value < 0.05 means the correlation is statistically significant
for param in ['RAM_GB', 'CPU_frequency', 'Storage_GB_SSD', 'Screen_Size_inch', 
              'Weight_pounds', 'CPU_core', 'OS', 'GPU', 'Category']:
    pearson_coef, p_value = stats.pearsonr(df[param], df['Price'])
    print(param)
    print("The Pearson Correlation Coefficient for ", param, " is", pearson_coef, " with a P-value of P =", p_value)
````
