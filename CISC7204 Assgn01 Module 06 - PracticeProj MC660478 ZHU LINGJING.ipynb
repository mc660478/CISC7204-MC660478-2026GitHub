Handle missing data:

- For continuous attributes (e.g., age), replace missing values with the mean.
- For categorical attributes (e.g., smoker), replace missing values with the most frequent value.
- Update the data types of the respective columns.
- Verify the update using `df.info()`.


````python
# For categorical attribute 'smoker', find the most frequent value (mode)
# value_counts().idxmax() returns the category that appears most often
is_smoker = df['smoker'].value_counts().idxmax()

# Replace missing values in 'smoker' with the most frequent entry
# Mode imputation is appropriate for categorical variables
df["smoker"].replace(np.nan, is_smoker, inplace=True)

# For continuous attribute 'age', calculate the mean age
# astype('float') ensures numeric calculation even if the column was read as string
mean_age = df['age'].astype('float').mean(axis=0)

# Replace missing values in 'age' with the calculated mean
# Mean imputation is suitable for continuous numerical variables
df["age"].replace(np.nan, mean_age, inplace=True)

# Update data types of 'age' and 'smoker' to integer for proper numeric representation
df[["age", "smoker"]] = df[["age", "smoker"]].astype("int")

# Verify the data type updates and confirm no missing values remain
print(df.info())
````

# Task 4 : Model Development

Fit a linear regression model that may be used to predict the `charges` value, just by using the `smoker` attribute of the dataset. Print the $ R^2 $ score of this model.


````python
# Import LinearRegression from sklearn for building regression models
from sklearn.linear_model import LinearRegression

# Define feature matrix X using only the 'smoker' column (single variable prediction)
# Double brackets [[ ]] create a 2D DataFrame as required by sklearn estimators
X = df[['smoker']]

# Define target vector Y with the 'charges' column (the value we want to predict)
Y = df['charges']

# Create a LinearRegression model instance
lm = LinearRegression()

# Fit the model to the data (train the model to learn the relationship)
lm.fit(X, Y)

# Calculate and print the R-squared score (coefficient of determination)
# R^2 measures how much variance in charges is explained by smoker status alone
print("R^2 score (smoker only):", lm.score(X, Y))
````

Fit a linear regression model that may be used to predict the `charges` value, just by using all other attributes of the dataset. Print the $ R^2 $ score of this model. You should see an improvement in the performance.


````python
# Define feature matrix Z using all attributes except charges (multivariate prediction)
# These are all the independent variables available in the dataset
Z = df[["age", "gender", "bmi", "no_of_children", "smoker", "region"]]

# Reuse the same LinearRegression instance (Y remains the same target variable)
# Fit the model using all available features to predict charges
lm.fit(Z, Y)

# Calculate and print the R-squared score for the multivariate model
# This should be higher than the single-variable model because more information is used
print("R^2 score (all features):", lm.score(Z, Y))
````

Create a training pipeline that uses `StandardScaler()`, `PolynomialFeatures()` and `LinearRegression()` to create a model that can predict the `charges` value using all the other attributes of the dataset. There should be even further improvement in the performance.


````python
# Import necessary components for building a machine learning pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.metrics import r2_score

# Define a pipeline with three sequential steps:
# 1. 'scale': StandardScaler normalizes features to zero mean and unit variance
# 2. 'polynomial': PolynomialFeatures creates interaction terms between features (degree=2 by default)
# 3. 'model': LinearRegression fits the final model on transformed features
Input = [('scale', StandardScaler()), 
         ('polynomial', PolynomialFeatures(include_bias=False)), 
         ('model', LinearRegression())]

# Create the Pipeline object with the defined steps
pipe = Pipeline(Input)

# Ensure Z is in float format for proper scaling and polynomial transformation
Z = Z.astype(float)

# Fit the pipeline on the feature matrix Z and target vector Y
pipe.fit(Z, Y)

# Predict charges using the fitted pipeline on the same data
ypipe = pipe.predict(Z)

# Calculate R^2 score by comparing actual Y values with predicted ypipe values
print("R^2 score (Pipeline):", r2_score(Y, ypipe))
````

# Task 5 : Model Refinement

Split the data into training and testing subsets, assuming that 20% of the data will be reserved for testing.


````python
# Import train_test_split for splitting data into training and testing subsets
from sklearn.model_selection import train_test_split

# Split the data into training (80%) and testing (20%) subsets
# random_state=1 ensures the split is reproducible across multiple runs
x_train, x_test, y_train, y_test = train_test_split(Z, Y, test_size=0.2, random_state=1)

# Print the number of samples in each subset to confirm the split
print("Number of test samples:", x_test.shape[0])
print("Number of training samples:", x_train.shape[0])
````

Initialize a Ridge regressor that used hyperparameter $ \alpha = 0.1 $. Fit the model using training data data subset. Print the $ R^2 $ score for the testing data.


````python
# Import Ridge regression which uses L2 regularization to prevent overfitting
from sklearn.linear_model import Ridge

# Create a Ridge regressor with regularization parameter alpha=0.1
# Alpha controls the strength of regularization (higher = more penalty on large coefficients)
RidgeModel = Ridge(alpha=0.1)

# Fit the Ridge model on the training data
RidgeModel.fit(x_train, y_train)

# Predict insurance charges for the test data
yhat = RidgeModel.predict(x_test)

# Calculate and print R^2 score on the test data to evaluate generalization performance
print("R^2 score (Ridge on test):", r2_score(y_test, yhat))
````

Apply polynomial transformation to the training parameters with degree=2. Use this transformed feature set to fit the same regression model, as above, using the training subset. Print the $ R^2 $ score for the testing subset.


````python
# Create PolynomialFeatures with degree=2 to generate squared terms and interaction terms
pr = PolynomialFeatures(degree=2)

# Fit and transform the training data to polynomial features
# fit_transform learns the feature mapping from training data and applies it
x_train_pr = pr.fit_transform(x_train)

# Transform the test data using the SAME mapping learned from training data
# We use transform (NOT fit_transform) to avoid data leakage from test set to training set
x_test_pr = pr.transform(x_test)

# Fit the Ridge model on the polynomial-transformed training data
RidgeModel.fit(x_train_pr, y_train)

# Predict on the polynomial-transformed test data
y_hat = RidgeModel.predict(x_test_pr)

# Calculate and print the R^2 score for the polynomial + Ridge model on test data
print("R^2 score (Polynomial + Ridge on test):", r2_score(y_test, y_hat))
````
